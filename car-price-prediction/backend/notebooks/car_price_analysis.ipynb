```python
plt.figure(figsize=(12, 6))

sns.countplot(
    data=df,
    x="company",
    order=df["company"].value_counts().index
)

plt.title("Number of Cars by Company")
plt.xlabel("Company")
plt.ylabel("Number of Cars")
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.countplot(
    data=df,
    x="fuel_type"
)

plt.title("Distribution of Fuel Types")
plt.xlabel("Fuel Type")
plt.ylabel("Number of Cars")

plt.show()
```

```python
eda_df = df.copy()

eda_df["price_numeric"] = (
    eda_df["Price"]
    .str.replace(",", "", regex=False)
    .replace("Ask For Price", np.nan)
    .astype(float)
)
```

```python
eda_df["price_numeric"].describe()
```

```python
plt.figure(figsize=(10, 6))

sns.histplot(
    data=eda_df,
    x="price_numeric",
    bins=40,
    kde=True
)

plt.title("Distribution of Car Prices")
plt.xlabel("Price")
plt.ylabel("Frequency")

plt.show()
```

```python
eda_df["year_numeric"] = pd.to_numeric(
    eda_df["year"],
    errors="coerce"
)
```

```python
plt.figure(figsize=(10, 6))

sns.histplot(
    data=eda_df,
    x="year_numeric",
    bins=20,
    kde=True
)

plt.title("Distribution of Manufacturing Year")
plt.xlabel("Year")
plt.ylabel("Number of Cars")

plt.show()
```
